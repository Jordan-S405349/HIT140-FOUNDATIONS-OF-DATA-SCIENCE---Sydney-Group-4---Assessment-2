```python
from pathlib import Path
from scipy import stats
import math
import pandas as pd
import matplotlib.pyplot as plt
```

```python
BASE = Path("__file__").resolve().parent.parent
PATH = BASE / "Datasets" / "task2_clean_sheets.csv"
OUT_DIR = Path("__file__").resolve().parent # saving output into script's own folder
```

```python
df = pd.read_csv(PATH)
sample = df.sample(40, random_state=42)
sample.to_csv(OUT_DIR / "analysis_sample_n40.csv", index=False)
x = sample["Clean_Sheets"]
```

```python
# Desciptive statistics
desc = pd.DataFrame({
    "n": [len(x)],
    "mean": [x.mean()],
    "median": [x.median()],
    "std": [x.std(ddof=1)],
    "min": [x.min()],
    "max": [x.max()]
})
desc.to_csv(OUT_DIR / "desciptive_statistics.csv", index=False)
```

```python
# Inferential statistics (one-sample t-test)
comparison_value = 1
alpha = 0.05
test = stats.ttest_1samp(x, comparison_value)
se = x.std(ddof=1) / (len(x) ** 0.5)
crit = stats.t.ppf(1 - alpha / 2, len(x) - 1)
ci_low, ci_high = x.mean() - crit * se, x.mean() + crit * se
decision = "Reject H0" if test.pvalue < alpha else "Fail to reject H0"

inferential = pd.DataFrame([{
    "test": "One-sample t-test",
    "comparison_value": comparison_value,
    "alpha": alpha,
    "n": len(x),
    "sample_mean": x.mean(),
    "standard_error": se,
    "t_statistic": test.statistic,
    "degrees_of_freedom": test.df,
    "p_value_two_sided": test.pvalue,
    "ci_95_low": ci_low,
    "ci_95_high": ci_high,
    "decision": decision,
}])
inferential.to_csv(OUT_DIR / "inferentioal_results.csv", index=False)
```

```python
# Summary
print("Task 2 - Clean Sheets")
print("Mean:", x.mean(), "Median:", x.median(), "SD:", x.std(ddof=1))
print("One-sample t:", test.statistic, "df:", test.df, "p:", test.pvalue)
print("95% CI:", (ci_low, ci_high))
print("Saved to:", OUT_DIR)
```
